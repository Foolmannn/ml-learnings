```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# Linear Regression
reg = LinearRegression()
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Ridge 
reg = Ridge(alpha=0.1)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Lasso
reg = Lasso(alpha=0.01)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# ElasticNet
reg = ElasticNet(alpha=0.005,l1_ratio=0.9) # this means 90 % weightage to the ridge and 10 % to the lasso 
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

So we can use the SGD Regressor For all the regularization with the penalty set to l1, l2 or elasticnet 

```python
from sklearn.linear_model import SGDRegressor
sgdreg = SGDRegressor(penalty='elasticnet',l1_ratio=0.9,max_iter=3000)
sgdreg.fit(X_train,y_train)
y_pred = sgdreg.predict(X_test)
r2_score(y_test,y_pred)
```
