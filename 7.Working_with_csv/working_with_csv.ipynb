Skip bad lines :
on_bad_lines : {'error', 'warn', 'skip'} or Callable, default 'error'
Specifies what to do upon encountering a bad line (a line with too many fields). Allowed values are :

'error', raise an Exception when a bad line is encountered.
'warn', raise a warning when a bad line is encountered and skip that line.
'skip', skip bad lines without raising or warning when they are encountered.

```python
pd.read_csv('bad.csv',on_bad_lines='skip') 
```

dtypes parameter


```python
pd.read_csv('bad.csv',on_bad_lines='skip').info() 
# here we have the float for the number of dependents which is not neccesary as it will just take more memory we can convert to int 
```

```python
pd.read_csv('bad.csv',on_bad_lines='skip',dtype={'NumberOfDependents':int}).info()

# now it will converted to integer type 
```

Handling Dates 

parse_dates : bool, list of Hashable, list of lists or dict of {Hashable
The behavior is as follows:

bool. If True -> try parsing the index. Note: Automatically set to True if date_format or date_parser arguments have been passed.
list of int or names. e.g. If [1, 2, 3] -> try parsing columns 1, 2, 3 each as a separate date column.
list of list. e.g. If [[1, 3]] -> combine columns 1 and 3 and parse as a single date column. Values are joined with a space before parsing.
dict, e.g. {'foo' ⁠:⁠ [1, 3]} -> parse columns 1, 3 as date and call result 'foo'. Values are joined with a space before parsing.
If a column or index cannot be represented as an array of datetime, say because of an unparsable value or a mixture of timezones, the column or index will be returned unaltered as an object data type. For non-standard datetime parsing, use ~pandas.to_datetime after ~pandas.read_csv.

```python
pd.read_csv('withdate.csv').info() # so date becomes the string by default. We cannot parse them as date 
```

```python
pd.read_csv('withdate.csv',parse_dates=['Month']).info() # now it becomes the datetime64 data type
```

```python
def removemonth(date):
    d =date.split(sep='-')
    return d[0]
removemonth("1948-01")
```

Convertors: 


```python
pd.read_csv('withdate.csv',converters={'Month': removemonth})
# this can be very helpful if we want some custom transformation
```
