```python
url = "https://jobejee.com/"

webpage = requests.get(url).text
# print(response)
```

```python
soup = BeautifulSoup(webpage, 'html.parser')
```

```python
print(soup.prettify())
```

```python
container = soup.find_all('div', class_="job-box-array")
```

```python
# print(container)

for data in container:
    title= data.find('a',class_="green-orange-link").text
    print(title)
    company=data.find('div',class_="sub-title").text
    print(company)
    location= data.find('div',class_="date").text
    print(location)
    salary= data.find('div',class_="salary").text
    print(salary)

```
