```python
jobTitle=[]
companyName=[]
address=[]
info=[]

for i in range(1,15):
    url=f"https://www.jobsnepal.com/jobs?page={i}"
    headers = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 Chrome/126.0 Safari/537.36"
        )
    }

    webpage = requests.get(url, headers=headers).text
    soup=BeautifulSoup(webpage,'html.parser')


    container = soup.find_all('div',class_="card-body")




    for data in container:
        title= data.find('h2',class_="job-title").text.strip()
        list = data.find_all('li')
        if len(title)<30 and len(list)==3:
            print(len(list))
            jobTitle.append(title)
            companyName.append(list[0].text.strip())
            address.append(list[1].text.strip())
            info.append(list[2].text.strip())
```

```python
print(jobTitle)
print(info)
```

Now form the data frame

```python
dic={'JobTitle':jobTitle,'CompanyName':companyName,'Location':address,'Responsibilities':info}
df = pd.DataFrame(dic)
```
