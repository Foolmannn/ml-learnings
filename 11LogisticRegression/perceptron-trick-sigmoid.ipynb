With sigmoid if the points is classified correctly it will push  and pull if incorrectly classified. 

Unlike with the step function which do nothing when they are correctly classified. 

And the further the distance from the model more agressively the pull or push will occur :
 
    So if the correctly classified then the closer point will push harder than the points far 
    
    So if the incorrectly classified then the farther point will pull harder than the points near  

```python
from sklearn.datasets import make_classification
import numpy as np
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=30)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron(X,y):
    
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1
    
    for i in range(1000):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]
        
    return weights[0],weights[1:]
        
```

With the step function 

```python
def step(z):
    return 1 if z>0 else 0
```

```python
intercept_,coef_ = perceptron(X,y)
```

```python
print(coef_)
print(intercept_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

As we can see that the builtin is still better than just using the sigmoid function . Although the sigmoid is better than the using simple step function 

So this is not the optimal solution. As we are chossing the random points for fixed epochs So we cannot guarantee that the solution is optimal .

Thus we will need the Metric to determine which solution is better ie LOSS FUNCTION 

So With Loss function : We can get the best model that performs the best 

```python
def perceptron(X,y):
    
    m = []
    b = []
    
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1
    
    for i in range(200):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]
        
        m.append(-(weights[1]/weights[2]))
        b.append(-(weights[0]/weights[2]))
        
    return m,b
```

```python
m,b = perceptron(X,y)
```
