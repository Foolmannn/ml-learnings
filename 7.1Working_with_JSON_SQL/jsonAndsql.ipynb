This will not work as I am using wsl and mysql is running in windows 

Yes, it will work, but "localhost" will not work in this setup. [1] 
When your Python code runs inside WSL, localhost refers to the WSL Linux environment itself, not your Windows host. Because your MySQL server is running on Windows, WSL cannot find it at that address. [2, 3, 4, 5] 
Follow these quick steps to make them talk to each other:
## 1. Get your Windows IP Address inside WSL
WSL automatically maps your Windows host IP. You can find it by running this command inside your WSL terminal: [6, 7] 

ip route show | grep default | awk '{print $3}'

(This will output an IP address, usually something like 172.25.80.1 or 192.168.x.x).
## 2. Update your Python Code
Replace localhost with the IP address you found in the step above:

conn = mysql.connector.connect(
    host="172.25.80.1",  # Replace with the exact IP from step 1
    user='root',
    password='root',
    database='world'
)

## 3. Allow Windows MySQL to Accept Remote Connections
By default, Windows MySQL only listens to requests coming from Windows itself. You need to tell it to listen to WSL. [8] 

   1. Open your MySQL Workbench or command line on Windows.
   2. Run this SQL command to see your root user restrictions:
   
   SELECT user, host FROM mysql.user WHERE user='root';
   
   3. If the host says localhost or 127.0.0.1, your Python code will still get blocked. You must allow connections from any host (%) by running:
   
   CREATE USER 'root'@'%' IDENTIFIED BY 'root';GRANT ALL PRIVILEGES ON *.* TO 'root'@'%' WITH GRANT OPTION;
   FLUSH PRIVILEGES;
   
   [9, 10, 11] 

## 4. Check Windows Firewall (If it still fails)
Windows Firewall might block WSL traffic on port 3306.

* Open Windows Defender Firewall on your PC.
* Click Advanced Settings -> Inbound Rules -> New Rule.
* Choose Port -> TCP -> Specific local ports: 3306.
* Select Allow the connection and save it as "MySQL WSL". [12, 13]
