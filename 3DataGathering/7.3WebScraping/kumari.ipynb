```python
import pandas as pd

import requests
import numpy as np

from bs4 import BeautifulSoup

url=f"https://www.kumarijob.com/"
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 Chrome/126.0 Safari/537.36"
    )
}

webpage = requests.get(url, headers=headers).text
soup=BeautifulSoup(webpage,'html.parser')


container = soup.find_all('div',id="job-grouping")
```

```python
# soup.prettify()
# print(container)
```

```python
# print(container)
len(container)
```

```python
data = container[0].find_all('div',class_='premium_card')
```

```python
len(data)
```

```python
jobTitle=[]
company=[]
location=[]
exp=[]
timeLeft=[]
salary=[]

for item in data:
    # i=item.find('a',class_='job-info').text
    # print(i)
    # j=item.find('a',title="View Company Details")
    # print(j)
    i=item.find_all('a',class_='job-info')

    # print(i)
    if len(i)>1:
        jobTitle.append(i[0].text.strip())
        # print(i[0].text)
        company.append(i[1].text.strip())
        # print(i[1].text)
        
        list = item.find_all('li')
        # print(list[0].text)
        if len(list)>3:
            location.append(list[0].text.strip())
            exp.append(list[1].text.strip())
            timeLeft.append(list[2].text.strip())
            salary.append(list[3].text.strip())
        else:
            location.append(list[0].text.strip())
            exp.append(list[1].text.strip())
            timeLeft.append(np.nan)
            salary.append(list[2].text.strip())
```

```python
# print(location)
# print(jobTitle)
print(timeLeft)
```

```python
dic={'JobTitle':jobTitle,'CompanyName':company,'Location':location,'Responsibilities':exp,'Deadline':timeLeft,"Salary":salary}
df = pd.DataFrame(dic)
# jobTitle=[]
# company=[]
# location=[]
# exp=[]
# timeLeft=[]
# salary=[]
```
