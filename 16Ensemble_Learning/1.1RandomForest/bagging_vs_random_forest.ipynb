#### Bagging is the general techniques where we can use the decicision tree, SVM, KNN as base model  where as Random Forest uses the bagging technique with only Decision Trees 

And another difference is in the way of the feature  selection (  Bootstraping ): 

In Bagging -> Tree level sampling is done ie The columns selected is before and only those columns will be used for the entire tree / model 

In Random Forest -> Node level sampling is done ie Columns selected is done on each node randomly ie for every node new random columns are selected ie more randomness 

```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.ensemble import BaggingClassifier,RandomForestClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
X,y = make_classification(n_features=5, n_redundant=0, n_informative=5,n_clusters_per_class=1)
```

```python
df = pd.DataFrame(X,columns=['col1','col2','col3','col4','col5'])
df['target'] = y
print(df.shape)
df.head()
```

```python
bag = BaggingClassifier(max_features=2)
```

```python
bag.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python
plt.figure(figsize=(12,12))
plot_tree(bag.estimators_[0])
plt.show()
```

### We can see only x[0] and x[1] which is sampled is being used on very decision node. So features is selected at Tree Level 

```python
rf = RandomForestClassifier(max_features=2)
```

```python
rf.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python
plt.figure(figsize=(12,12))
plot_tree(rf.estimators_[4])
plt.show()
```

### We can see x[0],x[1],x[3],x[2] and more are used  which is sampled on every decision node . So features is selected at Node Level ie More randomness 

### So Random forest beats the normal bagging generally due to more Randomness at every node level
