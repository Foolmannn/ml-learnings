```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['satisfaction','id']),df['satisfaction'],
                                                test_size=0.2)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler

# Create a small pipeline for numerical columns to handle missing values + scaling
numeric_transformer = Pipeline(
    steps=[
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler()),
    ]
)

transformer = ColumnTransformer(
    transformers=[
        (
            'num_scaled',
            numeric_transformer,
            [
                'Age',
                'Flight Distance',
                'Departure Delay in Minutes',
                'Arrival Delay in Minutes',
            ],
        ),
        (
            'o_encoded1',
            OrdinalEncoder(categories=[['disloyal Customer', 'Loyal Customer']]),
            ['Customer Type'],
        ),
        (
            'o_encoded2',
            OrdinalEncoder(categories=[['Eco', 'Eco Plus', 'Business']]),
            ['Class'],
        ),
        (
            'ohe',
            OneHotEncoder(sparse_output=False, drop='first'),
            ['Gender', 'Type of Travel'],
        ),
    ],
    remainder='passthrough',  # This will now safely catch your 0-5 survey rating columns
)

```

```python
Xtrain_new=transformer.fit_transform(X_train)
```

```python
Xtest_new=transformer.transform(X_test)
```

```python
Xtrain_new
```

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
```

```python

rf = RandomForestClassifier()

lr = LogisticRegression()
```

```python
rf.fit(Xtrain_new,y_train)
y_pred = rf.predict(Xtest_new)
accuracy_score(y_test,y_pred)
```

```python
lr.fit(Xtrain_new,y_train)
y_pred = lr.predict(Xtest_new)
accuracy_score(y_test,y_pred)
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    RocCurveDisplay
)
```

```python
model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)
```

```python
model.fit(Xtrain_new, y_train)
```

```python
y_pred = model.predict(Xtest_new)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
from sklearn.model_selection import RandomizedSearchCV


# 1. A streamlined, high-impact parameter grid
params = {
    'n_estimators': [150, 250, 400],
    'learning_rate': [0.03, 0.08, 0.15],
    'max_depth': [4, 6, 8],
    'subsample': [0.7, 0.9, 1.0],
    'colsample_bytree': [0.7, 0.9, 1.0],
    'reg_alpha': [0, 1],
    'reg_lambda': [1, 5],
}

# 2. Set up RandomizedSearchCV (testing only 25 random combinations)
random_search = RandomizedSearchCV(
    estimator=XGBClassifier(
        random_state=42, eval_metric='logloss', n_jobs=-1
    ),
    param_distributions=params,
    n_iter=25,  # Only evaluates 25 models instead of 1,296
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    random_state=42,
)

# 3. Fit the model
print('Starting hyperparameter tuning...')
random_search.fit(Xtrain_new, y_train)

# 4. View results
print('\nBest Parameters Found:')
print(random_search.best_params_)
print(f'Best Cross-Validation Accuracy: {random_search.best_score_:.4f}')

# 5. Extract the best model for predictions
best_model = random_search.best_estimator_
```
