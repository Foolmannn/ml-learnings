```python
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
encoder = LabelEncoder()
df['species'] = encoder.fit_transform(df['species'])
```

```python
df.head()
```

```python
df = df[['sepal_length','petal_length','species']]
```

```python
df.head()
```

```python
X = df.iloc[:,0:2]
y = df.iloc[:,-1]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# clf = LogisticRegression(multi_class='multinomial')
clf = LogisticRegression()
```

The code clf = LogisticRegression(multi_class='multinomial') fails because the multi_class parameter has been removed from scikit-learn (starting in version 1.7/1.8). [1] (https://app.semanticdiff.com/gh/google/jaxopt/pull/638/overview), [2] (https://scikit-learn.org/stable/whats_new/v1.5.html)

scikit-learn now automatically handles multiclass classification. If your target data has 3 or more classes, it automatically defaults to using the multinomial loss function

```python
clf.fit(X_train,y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
pd.DataFrame(confusion_matrix(y_test,y_pred))
```

```python
# prediction
query = np.array([[3.4,2.7]])
clf.predict_proba(query)
```

```python
clf.predict(query)
```

```python
from mlxtend.plotting import plot_decision_regions

plot_decision_regions(X.values, y.values, clf, legend=2)

# Adding axes annotations
plt.xlabel('sepal length [cm]')
plt.xlabel('petal length [cm]')
plt.title('Softmax on Iris')

plt.show()
```
