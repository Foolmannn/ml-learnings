The behavior changed across pandas versions, and that's why you'll see different tutorials online.

### Summary

| Pandas Version                                                          | `df.corr()` behavior                                                                                                                           |
| ----------------------------------------------------------------------- | ---------------------------------------------------------------------------------------------------------------------------------------------- |
| **1.5.x and earlier**                                                   | By default, non-numeric columns were ignored (`numeric_only=True` behavior).                                                                   |
| **2.0.x**                                                               | The default changed to `numeric_only=False`, so `df.corr()` tries to use all columns and raises an error if object/string columns are present. |
| **2.1.x, 2.2.x, 2.3.x, and newer (including 3.x development versions)** | Same behavior: you should explicitly specify `numeric_only=True` or select numeric columns yourself.                                           |

So:

* **If you're using pandas 1.5.x or older**, this works:

```python
df.corr()
```

* **If you're using pandas 2.0 or later**, you should write:

```python
df.corr(numeric_only=True)
```

or

```python
df.select_dtypes(include="number").corr()
```

---

### Why did pandas change this?

Before pandas 2.0:

```python
df.corr()
```

silently ignored all non-numeric columns.

The pandas developers changed the default to make the API **more explicit and consistent**. Instead of silently dropping columns, it now tries to use everything unless you tell it otherwise. This helps catch situations where users mistakenly think all columns are being analyzed.

### Best practice (recommended for all modern code)

Even if you're using an older version, many practitioners now prefer:

```python
corr = df.select_dtypes(include="number").corr()
```

This makes your intent explicit, works across pandas versions, and avoids surprises if your dataset contains string or categorical columns.
